```python
import pandas as pd
import requests
from io import StringIO

# Amazon Prime dataset URL
url = "https://raw.githubusercontent.com/ishagoel840/Amazon-Prime-Data-PowerBI/main/amazon_prime_titles.csv"

# Fetch CSV content safely
response = requests.get(url, verify=False)

# Load into pandas DataFrame
df = pd.read_csv(StringIO(response.text))

# Clean up duration column: extract numeric minutes or seasons
def parse_duration(x):
    try:
        if "min" in str(x):
            return int(x.replace("min", "").strip())
        elif "Season" in str(x):
            return int(x.split()[0]) * 60  # rough conversion: 1 season ≈ 60 mins
        else:
            return None
    except:
        return None

df['duration_num'] = df['duration'].apply(parse_duration)

# Aggregate data by country and release_year
aggregated_data = df.groupby(['country', 'release_year']).agg(
    title_count=('title', 'count'),
    avg_duration=('duration_num', 'mean')
)

print(aggregated_data.head(20))  # show first 20 rows
```

Merge

```python
import pandas as pd

# Create DataFrame for Dataset 1
data1 = [[1, 'Raj', 25], 
         [2, 'Sharma', 42], 
         [3, 'Saroj', 22], 
         [4, 'Raja', 51], 
         [5, 'Kajal', 26]]
df1 = pd.DataFrame(data1, columns=['ID', 'Name', 'Age'])

# Create DataFrame for Dataset 2
data2 = [[1, 'Male', 25000], 
         [2, 'Male', 42500], 
         [3, 'Female', 22300], 
         [4, 'Male', 51400], 
         [5, 'Female', 26800]]
df2 = pd.DataFrame(data2, columns=['ID', 'Gender', 'Salary'])
```

Inner merge

```python
merged_df = pd.merge(df1, df2, on='ID', how='inner')
print(merged_df)
```

Left Merge

```python
merged_df = pd.merge(df1, df2, on='ID', how='left')
print(merged_df)
```

Right Merge


```python
merged_df = pd.merge(df1, df2, on='ID', how='right')
print(merged_df)
```

Outet Merge

```python
merged_df = pd.merge(df1, df2, on='ID', how='outer')
print(merged_df)
```

Merging on Multiple

```python
df1['Country'] = ['USA', 'Canada', 'USA', 'Canada', 'USA']
df2['Country'] = ['USA', 'Canada', 'USA', 'Canada', 'Mexico']

# Merge on 'ID' and 'Country'
merged_df = pd.merge(df1, df2, on=['ID', 'Country'], how='inner')
print(merged_df)
```

Merging with different coloun names


```python
df2.rename(columns={'ID': 'EmployeeID'}, inplace=True)

# Merge using different column names
merged_df = pd.merge(df1, df2, left_on='ID', right_on='EmployeeID', how='inner')
print(merged_df)
```
