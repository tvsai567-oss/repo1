```python
import pandas as pd
data = {
    'customer_id': [1, 2, 3],
    'customer_name': ['Alice', 'Bob', 'Charlie'],
    'product_id': [101, 102, 103],
    'product_name': ['Widget', 'Gadget', 'Thingamajig'],
    'product_price': [19.99, 29.99, 39.99],
    'purchase_date': pd.to_datetime(['2024-01-01', '2024-01-02', '2024-01-03'])
}
df = pd.DataFrame(data)
print(df.select_dtypes(include='number'))
print(df.select_dtypes(include='object'))
print(df.select_dtypes(include='datetime'))
print(df.select_dtypes(exclude='object'))
print(df.dtypes)
```
